```python
import os

def get_size(path):
    total = 0

    if os.path.isfile(path):
        total = os.path.getsize(path)
    else:
        for root, dirs, files in os.walk(path):
            for file in files:
                total += os.path.getsize(os.path.join(root, file))

    return total

size_bytes = get_size(path)
size_gb = size_bytes / (1024 ** 3)

print(f"Dataset size: {size_gb:.2f} GB")
```

```python
from huggingface_hub import list_repo_files

repo_id = "ggaddam/OHLCV-1m"

files = list_repo_files(
    repo_id=repo_id,
    repo_type="dataset"
)

parquet_files = [
    f for f in files
    if f.endswith(".parquet")
]

print(f"Number of parquet files: {len(parquet_files)}")

for f in parquet_files:
    print(f)
```

```python
from huggingface_hub import HfApi

api = HfApi()

for f in parquet_files[:10]:
    try:
        metadata = api.get_paths_info(
            repo_id=repo_id,
            paths=f,
            repo_type="dataset"
        )

        print(f, metadata[0].size)

    except Exception as e:
        print(f, "ERROR:", e)
```

```python
import requests

base_url = f"https://huggingface.co/datasets/{repo_id}/resolve/main/"

file_sizes = {}

for f in parquet_files:
    url = base_url + f

    response = requests.head(
        url,
        allow_redirects=True
    )

    size = response.headers.get("Content-Length")

    if size:
        size_gb = int(size) / (1024**3)
        file_sizes[f] = size_gb
        print(f"{f}: {size_gb:.3f} GB")
    else:
        print(f"{f}: size unavailable")
```

```python
from huggingface_hub import list_repo_files

repo_id = "ggaddam/OHLCV-1m"

files = list_repo_files(
    repo_id=repo_id,
    repo_type="dataset"
)

parquet_files = sorted([
    f for f in files
    if f.startswith("data/ohlcv_") and f.endswith(".parquet")
], reverse=True)

print("Latest:", parquet_files[0])
print("Earliest:", parquet_files[-1])
print("Number of months:", len(parquet_files))
```

```python
target_gb = 13
selected_files = []
total_gb = 0

for f in parquet_files:
  size_f = file_sizes[f]

  if total_gb + size_f <= target_gb:
    selected_files.append(f)
    total_gb += size_gb
  else:
    break

print(f"# Files: {len(selected_files)}")
print(f"Total GB: {total_gb:.2f} GB")

print("Latest:", selected_files[0])
print("Earliest:", selected_files[-1])
```
