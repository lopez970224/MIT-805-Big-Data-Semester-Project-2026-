```python
from huggingface_hub import list_repo_files

repo_id = "ggaddam/OHLCV-1m"

files = list_repo_files(
    repo_id=repo_id,
    repo_type="dataset"
)

parquet_files = [
    f for f in files
    if f.endswith(".parquet")
]

print(f"Number of parquet files: {len(parquet_files)}")

for f in parquet_files:
    print(f)
```

```python
import requests

base_url = f"https://huggingface.co/datasets/{repo_id}/resolve/main/"

file_sizes = {}

for f in parquet_files:
    url = base_url + f

    response = requests.head(
        url,
        allow_redirects=True
    )

    size = response.headers.get("Content-Length")

    if size:
        size_gb = int(size) / (1024**3)
        file_sizes[f] = size_gb
        print(f"{f}: {size_gb:.3f} GB")
    else:
        print(f"{f}: size unavailable")
```

```python
from huggingface_hub import list_repo_files

repo_id = "ggaddam/OHLCV-1m"

files = list_repo_files(
    repo_id=repo_id,
    repo_type="dataset"
)

parquet_files = sorted([
    f for f in files
    if f.startswith("data/ohlcv_") and f.endswith(".parquet")
], reverse=True)

#print("Latest:", parquet_files[0])
#print("Earliest:", parquet_files[-1])
#print("Number of months:", len(parquet_files))
```

```python
target_gb = 30
selected_files = []
total_gb = 0

for f in parquet_files:
  size_f = file_sizes[f]

  if total_gb + size_f <= target_gb:
    selected_files.append(f)
    total_gb += size_gb
  else:
    break

print(f"# Files: {len(selected_files)}")
print(f"Total GB: {total_gb:.2f} GB")

print("Latest:", selected_files[0])
print("Earliest:", selected_files[-1])
```

**Observations:**

- The raw data range ranges between 2018-09 and 2025-07.

- Total size of the raw data is 30.13 GB

- Raw data is made up of 83 files.

**1.2. Downloading the working dataset**

```python
target_gb = 13
selected_files = []
total_gb = 0

for f in parquet_files:
  size_f = file_sizes[f]

  if total_gb + size_f <= target_gb:
    selected_files.append(f)
    total_gb += size_gb
  else:
    break

print(f"# Files: {len(selected_files)}")
print(f"Total GB: {total_gb:.2f} GB")

print("Latest:", selected_files[0])
print("Earliest:", selected_files[-1])
```

**Observations:**

- The working dataset size is 12.7 GB.

- The working dataset ranges from 09-20222 to 07-2025.

- The working contains 35 files.

```python
from huggingface_hub import hf_hub_download

repo_id = "ggaddam/OHLCV-1m"


for i in range(len(selected_files)):
  file_path = hf_hub_download(
      repo_id=repo_id,
      filename=selected_files[i],
      repo_type="dataset"
  )

  print(file_path)
```

**1.3. Displaying the working dataset into a dataframe**

```python
import pandas as pd

df = pd.read_parquet(file_path)
print("Shape:", df.shape)
display(df.head())
```

**Observations:**

- There's approximately 27 millions lines in the working dataset with 7 columns
- The dataset represent a 1 min OHLCV (Opening, High, Low, Closing and Volume)
